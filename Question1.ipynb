Case study
An ML team receives training records from several internal systems. Each record should contain a learner identifier and an engagement score, but scores arrive in different types and identifiers sometimes contain unnecessary spaces. Incomplete records may contain None. Before these records enter a training dataset, the team needs predictable cleaned data and a way to identify rejected records.

Coding task

Required function or decorator

INTERFACE

clean_training_records(records)

•	Accept learner_id only when it is a string that remains non-empty after trimming surrounding whitespace.

•	Accept score as an integer, float, or string convertible to a finite float. Reject Booleans, complex numbers, None, and other unsupported types. Accept only scores from 0 through 100.

•	Return accepted records with a trimmed learner_id, a float score, and an eligible Boolean that is True when score is at least 60.

•	Return rejected record positions using zero-based indexing. Preserve accepted-record order, retain repeated valid records, and ignore extra fields.

Sample input

PYTHON

    records = [{"learner_id": " L01 ", "score": "82.5"},
    {"learner_id": "L02", "score": 60},
    {"learner_id": "L03", "score": None},
    {"learner_id": "   ", "score": 75},
    {"learner_id": "L04", "score": True},
    {"learner_id": "L05", "score": 48.0},
    {"learner_id": "L06", "score": 105},]
    clean_training_records(records)


Question 1 continued

Expected output

EXPECTED RESULT


    {"cleaned": [
        {"learner_id": "L01", "score": 82.5, "eligible": True},
        {"learner_id": "L02", "score": 60.0, "eligible": True},
        {"learner_id": "L05", "score": 48.0, "eligible": False},
    ],
    "rejected_indices": [2, 3, 4, 6],}


Constraints and edge cases

  •	The outer input is a list. Reject elements that are not dictionaries and records missing required keys.
  
  •	Strings such as "nan" and "inf" are invalid because their converted values are not finite.
  
  •	An empty input returns empty cleaned and rejected_indices lists.

•	Input records must remain unchanged.

Working code, visible sample output, at least three additional meaningful test cases, and a brief explanation of your approach.



```python
import math

def clean_training_records(records):

    cleaned = []
    rejected_indices = []

    for i, record in enumerate(records):

        # Check that record is a dictionary
        if not isinstance(record, dict):

            rejected_indices.append(i)
            continue

        # Check required keys
        if "learner_id" not in record or "score" not in record:
            rejected_indices.append(i)
            continue

        learner_id = record["learner_id"]
        score = record["score"]

        # learner_id must be a string
        if not isinstance(learner_id, str):
            rejected_indices.append(i)
            continue

        # Remove surrounding spaces
        learner_id = learner_id.strip()

        # learner_id cannot be empty
        if learner_id == "":
            rejected_indices.append(i)
            continue

        # Boolean must be rejected
        if isinstance(score, bool):
            rejected_indices.append(i)
            continue

        # Score must be int, float, or string
        if not isinstance(score, (int, float, str)):
            rejected_indices.append(i)
            continue

        # Convert score to float
        try:
            score = float(score)
        except (ValueError, TypeError):
            rejected_indices.append(i)
            continue

        # Reject nan and infinity
        if not math.isfinite(score):
            rejected_indices.append(i)
            continue

        # Score must be between 0 and 100
        if score < 0 or score > 100:
            rejected_indices.append(i)
            continue

        # Add valid record
        cleaned.append({
            "learner_id": learner_id,
            "score": score,
            "eligible": score >= 60
        })

    return {
        "cleaned": cleaned,
        "rejected_indices": rejected_indices
    }
```

```python
records = [
    {"learner_id": " L01 ", "score": "82.5"},
    {"learner_id": "L02", "score": 60},
    {"learner_id": "L03", "score": None},
    {"learner_id": " ", "score": 75},
    {"learner_id": "L04", "score": True},
    {"learner_id": "L05", "score": 48.0},
    {"learner_id": "L06", "score": 105},
]
```

```python
from pprint import pprint

result = clean_training_records(records)

pprint(result)
```

* The function processes each training record one by one.
* First, it checks whether the record is a dictionary and contains the required `learner_id` and `score` fields.
* It then validates and cleans the learner ID by checking its type, removing surrounding spaces, and rejecting empty values.
* The score is checked to reject Boolean and unsupported types, converted to a float, and validated to ensure it is finite and between 0 and 100.
* Valid records are added to the cleaned list, and their eligibility is determined using `score >= 60`. Invalid records are recorded using their original index in `rejected_indices`.
